## Generating lc + Spectra

```python
#Define the function to generate light curves from, the default parameters output ~450-550 Supernovae.

def generate_lc(area = 4., tmin = 56075, tmax= 56425, zmax = 0.8, skynoise_min = 0.5, skynoise_max = 1,
                dust = sncosmo.CCM89Dust(), n_points = 300):
    
    spectra_t0 = []
    #Initialize the model
    model = sncosmo.Model(source='salt3-nir',
                      effects=[dust], 
                      effect_names=['host'], 
                      effect_frames=['rest']) 
    phases = np.random.uniform(tmin,tmax, size=n_points)
    #Observation parameters
    obs = Table({'time': phases,
             'band': ['lsstg', 'lsstr', 'lssti', 'lsstz', 'lssty']*(int(n_points/5)),
             'gain': [1.] * n_points,
             'skynoise': np.random.uniform(skynoise_min, skynoise_max, size=n_points),
             'zp': [28.]*n_points, 
             'zpsys':['ab']*n_points})

    """Initiate 1000 to 11000 array and zero out the waves under 2000 when simulating and zero weight (infinite uncertainty)"""
    
    """Generate all possible set of wavelengths and set zero wave zero flux where we have no data"""
    #Get a redshift distribution
    redshifts = list(sncosmo.zdist(0., zmax, time=(tmax-tmin), area=area))
    params = []
    s2n_max = 20 
    print(f' These distributions output {len(redshifts)} SNe')
    #Loop over redshifts to get amplitudes
    for z in redshifts:
        mabs = np.random.normal(-19.3, 0.3)
        model.set(z=z)
        model.set_source_peakabsmag(mabs, 'lsstr', 'ab')
        x0 = model.get('x0')
        t0 = np.random.uniform(tmin, tmax)
        #Set Parameters
        p = {'z':z, 't0':t0, 'x0':x0, 'x1': np.random.normal(0., 1.), 'c': np.random.normal(0.,0.1)}
        params.append(p)
        model.set(**p)
        
        #Convert rest-frame wavelengths to observer-frame wavelengths and zero the spectrum in the regions we don't
        #get data for
        z_waves = waves * (1+z)
        z_waves_sub2 = z_waves[z_waves<4000]
        z_waves_salt = z_waves[z_waves >=4000]
        spec_t0 = np.zeros((n_points, len(z_waves_sub2)))
        spec_t0_salt = model.flux(phases, z_waves_salt)
        spec_t0 = np.append(spec_t0, spec_t0_salt, axis=1)
        spec_t0 = [i*(10**20) for i in spec_t0]
        fluxerr_t0 = []
        #weights_t0 = []
        spec_err_t0 = []
        for spec in spec_t0:
            fluxerr = []
            specerr = []
            #weights = []
            for spec_point in spec:
                if spec_point == 0:
                    fluxerr.append(np.inf)
                    specerr.append(0.)
                else:
                    fluxerr.append(np.abs(spec_point/s2n_max))
                    specerr.append(np.abs(spec_point/s2n_max) * np.random.normal())
            fluxerr_t0.append(fluxerr)
            spec_err_t0.append(specerr)
        #Initialize a table with the spectra elements.
        spec_obs = Table({'time': phases,
                          'spectrum': spec_t0,
                          'wavelengths': [z_waves]*n_points,
                          'SN redshift': [z]*n_points,
                          'fluxerr': fluxerr_t0,
                          'spec_err': spec_err_t0,
                         })
        spec_obs['spectrum'] = spec_obs['spectrum'] + spec_obs['spec_err']
        spectra_t0.append(spec_obs)

    #Get light curve data and convert to dataset
    lcs = sncosmo.realize_lcs(obs, model, params)
    dataset = lcdata.from_light_curves(lcs)
    #Adding spectra into list of lists for each supernovae
    spectra = spectra_t0
    return dataset, spectra
light_curve, spectra = generate_lc()
```

```python
#Use model preprocessing on ligh curves
dataset = model.preprocess(light_curve)
```

```python
def preprocess_spectra(spectra):
    """Function to preprocess the observed spectra to use in the loss function calculations against the model spectra
    
    Parameters:
    -----------
    spectra: List of `astropy.Table`
    
    Returns:
    --------
    dict of `torch.Tensor` elements
    """
    
    from torch.nn.utils.rnn import pad_sequence
    
    obs_spectra_vals = []
    obs_spectra_bins = []
    obs_time = []
    obs_weights = []
    
    #Get data needed by ParSNIP for comparison
    data = model._get_data(dataset.light_curves)
    time_cp = data['compare_data'][:,0]
    
    #Loop and get ParSNIP grid time for each spectra and mask with compare_data, convert to tensor and append to list
    for idx, spectrum in enumerate(spectra):
        grid_time = parsnip.time_to_grid(spectrum['time'], dataset.meta[idx]['parsnip_reference_time'])
        grid_time = grid_time.data
        grid_time = np.array(grid_time, dtype=np.float64)
        grid_time = np.around(grid_time, 2)
        
        time_comp = time_cp[idx].cpu().detach().numpy()
        time_comp = np.array(time_comp, dtype=np.float64)
        time_comp = np.around(time_comp, 2)
        
        time_mask = np.isin(grid_time, time_comp)
        spectrum = spectrum[time_mask]
        
        time = spectrum['time']
        time = torch.FloatTensor(time)
        obs_time.append(time)
        
        spec = spectrum['spectrum']
        spec = np.array(spec, dtype=np.float64)
        spec = torch.from_numpy(spec)
        obs_spectra_vals.append(spec)
        
        bins = spectrum['wavelengths']
        bins = torch.tensor(bins)
        obs_spectra_bins.append(bins)
        
        #Calculate weights for each time point for each supernova
        weights = 1 / (spectrum['fluxerr']**2 + ((model.settings['error_floor']*spec.detach().numpy())**2))
        weights = torch.tensor(weights)
        obs_weights.append(weights)
        
    #Pad all tensors to match dimensions
    obs_time = pad_sequence(obs_time, batch_first = True)
    
    obs_spectra_vals = pad_sequence(obs_spectra_vals, batch_first = True)
    obs_spectra_vals = obs_spectra_vals.permute(0,2,1)
    
    obs_spectra_bins = pad_sequence(obs_spectra_vals, batch_first = True)
    obs_spectra_bins = obs_spectra_bins.permute(0,2,1)
    
    obs_weights = pad_sequence(obs_weights, batch_first = True)
    obs_weights = obs_weights.permute(0,2,1)
    
    obs_spectra = {
        'time': obs_time,
        'flux': obs_spectra_vals,
        'bins': obs_spectra_bins,
        'weights': obs_weights
    }
    
    return obs_spectra
spectra_prc = preprocess_spectra(spectra)
```

```python
weight = spectra_prc['weights']
obs_spectra = spectra_prc['flux']
model_spectra = result['model_spectra']
#weight = model.settings['error_floor']**2 * weight

num = torch.sum(weight * model_spectra * obs_spectra, dim=1)
denom = torch.sum(weight * model_spectra * model_spectra, dim=1)

denom[denom == 0.] = 1e-5

amplitude_mu = num/denom
amplitude_logvar = torch.log(1. / denom)

print(f'The tensor amplitude_mu has shape {amplitude_mu.shape} and is equal to {amplitude_mu}' )
print(f'The tensor amplitude_logvar has shape {amplitude_logvar.shape} and is equal to {amplitude_logvar}')
```

```python
amplitude = model._reparameterize(amplitude_mu, amplitude_logvar, sample=True)
model_spectra = model_spectra * amplitude[:, None]
```

```python
nll_sum = 0
for i in range(0,obs_spectra.shape[0]):
    max_idx = np.argmax(result['obs_flux'][i])
    nll = (0.5 * weight[i,:,max_idx]
               * (obs_spectra[i,:,max_idx] - model_spectra[i,:,max_idx])**2)
    nll_sum += nll
    plt.plot(model_spectra[i,:,max_idx].detach().numpy())
    plt.plot(obs_spectra[i,:,max_idx].detach().numpy())
    plt.plot(1 / np.sqrt((weight[i,:,max_idx].detach().numpy())))
    plt.title(f'The nll for Supernova no. {i} is {nll.sum()}')
    plt.show()
```
